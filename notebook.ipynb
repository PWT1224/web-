# 5，换行和水平线标签

## “br”换行“hr”水平线
- 浏览器无法识别代码终中的回车换行

```python
<!DOCTYPE html>
<html lang="en">
<head>
    <meta charset="UTF-8">
    <meta name="viewport" content="width=device-width, initial-scale=1.0">
    <title>Document</title>
</head>
<body>
    第一行内容
    <br>
    第二行内容
    <hr>
</body>
</html>
```

# 6，文本格式化标签

## 在下图中我们常使用左边的标签来进行编辑

```python
<!DOCTYPE html>
<html lang="en">

<head>
        <meta charset="UTF-8">
        <meta name="viewport" content="width=device-width, initial-scale=1.0">
        <title>Document</title>
</head>

<body>
        <br>
        <strong>加粗语句</strong>
        <br>
        <em>倾斜语句</em>
        <br>
        <ins>下划线语句</ins>
        <br>
        <del>删除线</del>
</body>

</html>
```

# 7，图像标签

## “img src = "图片的URL"”

```python
<!DOCTYPE html>
<html lang="en">
<head>
    <meta charset="UTF-8">
    <meta name="viewport" content="width=device-width, initial-scale=1.0">
    <title>Document</title>
</head>
<body>
    <img src="">
    <!-- 用"./直接进行选择图片" -->
</body>
</html>
```

## 图像标签的属性

- 注：照片的高以及其他属性我们在css文件上修改

```python
<!DOCTYPE html>
<html lang="en">

<head>
        <meta charset="UTF-8">
        <meta name="viewport" content="width=device-width, initial-scale=1.0">
        <title>Document</title>
</head>

<body>
        <img src="./1764999657354.jpg" alt='图片无法查看' title='帅照'>
        
        <img src="./1764999659573.jpg" alt="图片无法查看" title='帅照'>
        <img src="./1764999657354.jpg" width="100">
        <img src="./1764999657354.jpg" height="100">
</body>

</html> 
```
